```python
import requests
from PIL import Image
from io import BytesIO

transforms = [
    "1col",
    "2col",
    "3col",
    "4col",
    "6col",
    "8col",
    "12col",
    "16col",
]

base = "https://media.formula1.com/d_driver_fallback_image.png/content/dam/fom-website/drivers/L/LANNOR01_Lando_Norris/lannor01.png.transform/{}/image.png"

for t in transforms:
    try:
        url = base.format(t)
        r = requests.get(url, timeout=10)
        img = Image.open(BytesIO(r.content))
        print(f"{t:5} -> {img.size}  ({r.status_code})")
    except Exception as e:
        print(f"{t:5} -> ERROR: {e}")
```
